# ୨୧ ─── iris з котиком ─── ୨୧


## ♡ дані


```python
print("୨୧ ─── IRIS З КОТИКОМ ─── ୨୧")
print(r"       /\_/\ ")
print("      (｡•ᴗ•｡) ♡")
print("       > 🎀 <")
print()

from sklearn import datasets
from sklearn.model_selection import train_test_split
iris = datasets.load_iris()
train_data, test_data, train_labels, test_labels = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42
)
print("♡ Навчальних квіток:", len(train_data))
print("♡ Тестових квіток:", len(test_data))
print("✿ Назви класів:", iris.target_names)
```

## ♡ knn

беремо 5 найближчих квіток.


```python
k = 5
predictions = []
for test_item in test_data:
    distances = []
    for train_item in train_data:
        distance_squared = 0
        for x, y in zip(test_item, train_item):
            distance_squared += (x - y) ** 2
        distance = distance_squared ** 0.5
        distances.append(distance)
    sorted_neighbors = sorted(zip(distances, train_labels))
    nearest = sorted_neighbors[:k]
    count0 = 0
    count1 = 0
    count2 = 0
    for distance, label in nearest:
        if label == 0:
            count0 += 1
        elif label == 1:
            count1 += 1
        else:
            count2 += 1
    predicted_class = 0
    biggest_count = count0
    if count1 > biggest_count:
        biggest_count = count1
        predicted_class = 1
    if count2 > biggest_count:
        biggest_count = count2
        predicted_class = 2
    predictions.append(predicted_class)
print("୨୧ ─── ПРОГНОЗИ ─── ୨୧")
print("♡ Прогнози:       ", predictions)
print("♡ Правильні класи:", [int(label) for label in test_labels])
```

## ♡ accuracy

правильні прогнози / всі прогнози.


```python
correct = 0

for i in range(len(test_labels)):
    if predictions[i] == test_labels[i]:
        correct += 1

accuracy = correct / len(test_labels)
print("\n୨୧ ─── ACCURACY ─── ୨୧")
print("♡ Правильних прогнозів:", correct, "з", len(test_labels))
print("♡ Accuracy:", accuracy)
```

## ♡ tp, fp, fn, tn

кожен клас окремо.


```python
print("୨୧ ─── РАХУЄМО КЛАСИ ─── ୨୧")

counts = []

for class_id in range(3):
    tp = 0
    fp = 0
    fn = 0
    tn = 0

    for i in range(len(test_labels)):
        if test_labels[i] == class_id and predictions[i] == class_id:
            tp += 1
        elif test_labels[i] != class_id and predictions[i] == class_id:
            fp += 1
        elif test_labels[i] == class_id and predictions[i] != class_id:
            fn += 1
        else:
            tn += 1

    counts.append([tp, fp, fn, tn])
    print("\n✿ Клас:", class_id, "—", iris.target_names[class_id])
    print("  ♡ TP:", tp, "FP:", fp, "FN:", fn, "TN:", tn)
```

## ♡ precision, recall, f1

precision = tp / (tp + fp)

recall = tp / (tp + fn)

f1 = 2 * precision * recall / (precision + recall)


```python
print("୨୧ ─── РЕЗУЛЬТАТ ─── ୨୧")

for class_id in range(3):
    tp, fp, fn, tn = counts[class_id]

    precision = 0
    if tp + fp > 0:
        precision = tp / (tp + fp)

    recall = 0
    if tp + fn > 0:
        recall = tp / (tp + fn)

    f1 = 0
    if precision + recall > 0:
        f1 = 2 * precision * recall / (precision + recall)

    print("\n✿ Клас:", class_id, "—", iris.target_names[class_id])
    print("  ♡ Precision:", precision)
    print("  ♡ Recall:", recall)
    print("  ♡ F1:", f1)

print("\n୨୧ ───────────────────────── ୨୧")
print("     квіточки пораховано ♡")
print("        (｡•ᴗ•｡)")
print("୨୧ ───────────────────────── ୨୧")
```

## ♡ відповіді


1. precision — скільки вгадано серед прогнозів класу. recall — скільки знайдено серед справжніх.
2. так, якщо один рідкісний клас погано вгадується.
3. усі три однаково: f1 = 1.
